### 1. 리스트(List)
- 순서대로 나란히 줄 세운 값
- 대괄호 `[]` 사용
- 가변성. 데이터 추가, 삭제, 수정이 자유로움

```python
import random

random.seed(42)
my_list = random.sample(range(1, 21), 8)
my_list # 랜덤함수를 활용하여 임의 값 생성
print(f'리스트 : {my_list}')

# 인덱싱
print(f'5번째 값 : {my_list[4]}')

# 슬라이싱
print(f'1-4번째 값 슬라이싱 :  {my_list[0:4]}')

# 값 수정
my_list[4] = 30
print(f'수정된 리스트 : {my_list}')

# 값 삽입
my_list.append(24) # 단건 추가
my_list.extend([8, 26]) # 다건 추가
print(f'삽입된 리스트 {my_list}')

# 값 삭제
my_list.pop() # index로 삭제, 값 안 넣으면 맨 뒤에 값 삭제
my_list.remove(8) # 맨 처음 나온 8 삭제
print(f'삭제된 리스트 {my_list}')
```

### 2. 튜플 (Tuple)
- 소괄호 `()` 사용
- 불변성. 한 번 생성하면 안의 값 바꿀 수 없음.

```python
my_tuple = (1, 2, 3, 1, 2, 3)
print(my_tuple)

my_tuple[0] = 11 # 값 변환 시, TypeError 발생
```

### 3. 딕셔너리(Dictionary)
- key-value 값
- 수정/삭제 가능

```python
dic = {
    '수학' : [90, 95, 70],
    '영어' : [80, 88, 100],
    '국어' : [95, 88, 35]
}
dic['수학']
```

### 4. 데이터프레임 인덱싱/슬라이싱
- iloc(Integer location) : 숫자로 접근
- loc(location) : 라벨링으로 접근 (df.loc[행, 열])

```python
import pandas as pd

df = pd.DataFrame(dic)
df
```

```python
# 인덱싱
print(f'두번째 col값: {df.iloc[1]}')
print('1번 학생의 국어 점수 : ', df.loc[0, '국어'])
print('2번 학생의 과목별 점수 : ')
print(df.loc[1])
```

```python
# 슬라이싱
print('1~2번 학생의 영어, 국어 점수 :')
print(df.loc[0:1, '영어':'국어'])

print('1~2번 학생의 과목별 점수 :')
print(df.iloc[1:3]) # iloc의 경우 마지막 값 직전까지 슬라이싱
```
